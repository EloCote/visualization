Visualizations for advocacy

--> Rational appeal
--> Moral appeal
--> Emotional appeal

Activity:
Rational appeal: descriptions of the conditions, specific numbers
Moral appeal: highlight inhumane conditions (e.g., crawling, numbers of people)
Emotional appeal: evokes anger, makes us feel empathy

Two functions:
Presentation: describe facts
Representation: subjectively depciting ideas using metaphors, analogy allegory
--> most use both

Representation --> represent what is "below the surface", unseen contributions

```python
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
tips = sns.load_dataset('tips')
print(tips)
```

```python
sns.set_style('dark') #change style
sns.lineplot(data = tips, #chose dataset
x = 'total_bill', #define x variable
y= 'tip') #define y variable
```

```python
tipgraph = sns.lineplot(data=tips, 
                        x = 'total_bill',
                        y='tip')
tipgraph.set(title='Tips vs. Total Bill', ####Add titles
            xlabel ='Total Bill ($)',
            ylabel = 'Tip Amount ($)')
```

```python
#modifying easthetic elements
tipgraph = sns.lineplot(data=tips,
                        x = 'total_bill',
                        y='tip',
                        color='hotpink',
                        linestyle = '--',
                        linewidth =3,
                        marker = 'o',
                        markerfacecolor = 'indigo')
```

```python
#Add multiple variables
tipgraph = sns.lineplot(data=tips,
                        x = 'total_bill',
                        y='tip', style = 'time', hue='day'
                        ,palette = ['purple', 'hotpink', 'deepskyblue', 'yellowgreen'])

tipgraph.set(title='Tips vs. Total Bill',
            xlabel ='Total Bill ($)',
            ylabel = 'Tip Amount ($)')
```

```python
#Pair plot
sns.pairplot(
    data = tips,
    hue = 'day'
)
```

```python
#Relplot --> lets us explore specific levels within variables
daysplot = sns.relplot(data=tips,
                        x = 'total_bill',
                        y='tip',
                        hue = 'sex',
                        col = 'day',
                        kind = 'scatter',
                        col_wrap=2)
```

```python
import plotly.graph_objects as go
from wordcloud import WordCloud
from matplotlib_venn import venn2, venn2_circles, venn2_unweighted
```

```python
x1 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y1 = np.array([110, 180, 240, 99, 220])

graph=go.Figure()
graph.add_trace(go.Bar(x=x1, y=y1))

graph.update_layout(
    title="Pirates Scores",
    xaxis_title="Pirates",
    yaxis_title="Score"
)

graph.write_html("pirategraph.html")
```

```python
graph=go.Figure()
graph.add_trace(go.Scatter(x=x1, y=y1, mode="markers", marker =dict(size=15, color='hotpink', opacity=1, line=dict(width=5, color='purple'))))
graph.update_layout(title = "Interactive Pirate Plot",
xaxis_title = 'Pirates',
yaxis_title = "Scores",
width=500,
height=500)
```
