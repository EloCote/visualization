```python
import numpy as np 
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
np.random.seed(613) #randomly generate some variables
x = np.arange(50)
y1 = np.random.randint(0, 100, 50)
y2 = np.random.randint(0, 100, 50)
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1)
ax.plot(x,y2)
```

```python
##Add legend
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1, label='Person 1')
ax.plot(x,y2, label='Person 2')
ax.legend(loc='lower right',
          frameon = True, ##add frame around the legend
          fontsize = 6, #change font size
          ncol = 1, #specify number of columns
          shadow = True)
```

```python
##Moving legend outise of the plot area
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1, label='Person 1')
ax.plot(x,y2, label='Person 2')
ax.legend(loc='upper left', bbox_to_anchor=(1,1))
```

```python
##Annotations with the text() function --> specify coordiantes and text
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label='Person 1')
ax.scatter(x,y2, label='Person 2')
ax.legend(loc='lower right')
ax.text(10, 95, 'This value is important')
```

```python
##Modifying text annotations
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label='Person 1')
ax.scatter(x,y2, label='Person 2')
ax.legend(loc='lower right')
ax.text(10, 95, 'This value is important', 
        ha='left' #alignment of text
        , color='red' #modify font color
        ,size=20) #modify font size
```

```python
fig, ax = plt.subplots()
ax.axis([0,10,0,10])
ax.text(1,5, ".Data:(1,5)", transform=ax.transData) #refer to specific data point
ax.text(0.5, 0.1, ".Axes:(0.5, 0.1)", transform=ax.transAxes) #fraction referring to axes
ax.text(0.2, 0.2, ".Figure:(0.2, 0.1)", transform=fig.transFigure) #fraction reffering to panel
```

```python
##Annotating with arrows
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label='Person 1')
ax.scatter(x,y2, label='Person 2')
ax.legend(loc='lower right')
ax.annotate('This is important!', xy=(10,95), xytext=(20,70), arrowprops=dict(facecolor='black'))
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label='Person 1')
ax.scatter(x,y2, label='Person 2')
ax.legend(loc='lower right')
ax.annotate('This is important!', xy=(10,95), xytext=(20,70), arrowprops=dict(arrowstyle='wedge',color='hotpink'))
```

```python
#Removing tick marks and labels
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label='Person 1')
ax.scatter(x,y2, label='Person 2')
ax.legend(loc='lower right')
ax.yaxis.set_major_locator(plt.NullLocator()) #remove tick marks and labels on y axis
ax.xaxis.set_major_formatter(plt.NullFormatter()) #remove tick marks and labels on x axis
```

```python
#Limiting tick marks
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label='Person 1')
ax.scatter(x,y2, label='Person 2')
ax.legend(loc='lower right')
ax.xaxis.set_major_locator(plt.MaxNLocator(3))
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label='Person 1')
ax.scatter(x,y2, label='Person 2')
ax.legend(loc='lower right')
ax.xaxis.set_major_locator(plt.MultipleLocator(5)) #intervals of 5
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label='Person 1')
ax.scatter(x,y2, label='Person 2')
ax.legend(loc='lower right')
plt.xticks(rotation=45,ha='right') #rotate labels
```

```python
#Modify axis title fonts
font1 ={'family':'serif','color': 'indigo'}
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label='Person 1')
ax.scatter(x,y2, label='Person 2')
ax.legend(loc='lower right')
plt.xlabel('Shiny New X Axis!', fontsize = 18, fontdict = font1)
```
