```python
# Libraries
import numpy as np
import matplotlib.pyplot as plt
```

```python

# Make a random dataset:
height = [3, 12, 5, 18, 45]
bars = ('A', 'B', 'C', 'D', 'E')
y_pos = np.arange(len(bars))

# Create bars
plt.bar(y_pos, height)

# Create names on the x-axis
plt.xticks(y_pos, bars)

# Show graphic
plt.show()

##This plot is really clean and easy to understand. It's simple as there are only a few elements wich decreases cognitive load.
##It is easy to grasp the differences between the categories.
```

```python
# create data: an array of values
size_of_groups=[12,11,3,30]

# Create a pieplot
plt.pie(size_of_groups)
plt.show()

##This pie chart might be less effective as it is harder for people to clearly see the differences between the
##different groups compared to a bar graph as above.
```
