Accessibility of visualizations

Choice of colours --> can be not suitable for people with colour blindness
Unaccessible can be damaging to health --> e.g., tacile or braille versions of COVID-19 data visualizations are rare 

We have responsibility to priotize accessibility!

Should be clear concise, and easy to use
Present info in predictable, barrier-free ways
Provide barrier-free ways for people to interact with you
Need to work with assistive technologies devices

Using colour
lumination --> how bright, saturation --> how colourful = suitable for ordered data
hue --> pure colours = suitable for categorical data

small objects --> should have more saturation

Colour blindness simulator

overly bright colours can be painful to look at and overwhelming
higher contrast make data visualizations easier to tell apart

WebAIM --> check contrast

Viridis package --> make plots easier to read for colourblindness 

Text

Data labels can make it easier to follow
Every visual element should be described in legend

typeface --> set of text character
font --> subset of typeface

Sans serif fonts are more accessible for people with cognitive disabilities

should be at least 12 points size or greater
monospaced text is more accessible

Alt text --> make it accessuble to people using a screen reader or unable to see image
convey all essential info about visualization

Activity: A man is talking to a female scientist

Most appropriate alt-text depends on intended purpose of an image or visaulization

For data-centric images --> additional considerations 

Four types of desrciptive content

Activity: level 1: e.g., axis time of the day, level 2: e.g., descriptive stats such as describe average time people spend on each category, correlations etc. level 3: perceptual interpretation of data, e.g., people spend create time during day, level 4: routines depend on time period (subjective), interpretation that goes beyond data

four types are nt equally helpful

blind and sighted people found level 1 content on its won the least helpful

blind participants found level 4 the least useful

most useful include descriptions from multiple levels (2 and 3)

Can people actually access our data visualizations?

should be functional and visible in chrome, firefox, internet explorer, safari

Cost to making data visualization accessible

User testing + gathering feedback --> good way to make it accessible


```python
import numpy as np 
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
np.random.seed(613)
x1 = np.arange(50)
y1 = np.random.randint(0, 75, 50)
x2 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y2 = np.array([110, 180, 240, 99, 220])
```

```python
fig, (ax1, ax2) = plt.subplots(ncols=2, nrows=1, figsize=(7,3))
ax1.scatter(x1, y1)
ax2.bar(x2, y2)
```

```python
fig, (ax1, ax2) = plt.subplots(ncols=2, nrows=1, figsize=(7,3))
ax1.scatter(x1, y1)
ax1.annotate("this is important", xy=(10,60), xytext=(20,60), arrowprops=dict(facecolor='black'))
ax2.bar(x2,y2)
```

```python
#Subplots without grid arrangement
fig, someaxes = plt.subplot_mosaic([['ax1','ax3'],['ax2', 'ax3']],figsize=(7,4), layout='constrained') #layout='contrained' --> labels adjusted on size of subplots
someaxes["ax1"].scatter(x1, y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1)    

someaxes["ax1"].set_xlabel('A Big Label', fontsize=18)
someaxes["ax2"].set_xlabel('Another Label', fontsize=18)
someaxes["ax3"].set_xlabel('Label 2: 2 Fast 2 Furious', fontsize=18)
```
