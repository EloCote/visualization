```python
#Import packages

import numpy as np 
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
np.random.seed(123)
x = np.arange(50)
y = np.random.randint(0, 100, 50)#randomly generate dataset
```

```python
#basic scatterplot
fig, ax = plt.subplots(figsize=(5, 3)) #ax = subplot area where draw data points, width: 5 inches, height: 3 inches
ax.scatter(x, y)
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.bar(x,y)
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.hist(y) #histogram, only need y
```

```python
#Add labels
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)
ax.set_title('Total growth over time')
ax.set_ylabel('Total growth')
ax.set_xlabel('Years since start')
fig.tight_layout() #change width and height to make cleaner
```

```python
#Modify labels
font1={'family':'sans-serif', 'color': 'blue', 'size':20}
font2={'family':'monospace', 'color':'green', 'size':14}
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)
ax.set_title('Total growth over time', fontdict=font1,loc='left') #loc specifies how title aligns with margins of subplots
ax.set_ylabel('Total growth', fontdict=font2)
ax.set_xlabel('Years since start', fontdict=font2)
fig.tight_layout()
```

```python
#Customize plot appearance
fig, ax = plt.subplots(figsize=(5, 3)) 
ax.scatter(x,y, marker='*', color='indigo')
#fig.show() 
```

```python
#connect scatter points with line
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y,marker='*',color='indigo', linestyle='--', linewidth=2)
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y,marker='*',color='#7425b9', linestyle='--', linewidth=2) #modify colour with hex code
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y,marker='*',color='#7425b9', linestyle='--', linewidth=2, markersize=12, markeredgecolor='#fa9359', markerfacecolor='#000000') 
```

```python
#Grid lines
fig, ax = plt.subplots(figsize=(5, 3))
ax.grid(axis='y', color='blue', linewidth=2, linestyle='-.')
ax.plot(x,y,marker='*',color='#7425b9', linestyle='--', linewidth=2, markersize=12, markeredgecolor='#fa9359', markerfacecolor='#000000') 
```
