```python
#Import packages
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
#Download dataset
hate = pd.read_csv('hate_data.csv')
print(hate) #Print dataset
```

```python
#Count frequency of hate crimes for each year
# Source - https://stackoverflow.com/q/54789128
# Posted by daiyue
# Retrieved 2026-06-18, License - CC BY-SA 4.0
hate["frequency"] = (hate.groupby('OCCURRENCE_YEAR')['EVENT_UNIQUE_ID'].transform('nunique'))
```

```python
sns.set_style('dark') #Change style
hategraph = sns.lineplot(data=hate, ##Create line plot graph
                        x = 'REPORTED_YEAR', #X Variable : Year
                        y='frequency', #Y Variable : Frequency
                        color='blue', #Colour of line
                        linestyle = '-', #Style of line
                        linewidth =2, #Width of line
                        ) 
hategraph.set(title='Frequency of Hate Crimes per Year in Toronto (2018-2025)', ####Add title
            xlabel ='Year', ##X label
            ylabel = 'Frequency of Hate Crimes') ##Y label
```
